```python
from statsmodels.tsa.holtwinters import Holt
```

```python
df['Forecasting'] = pd.to_datetime(df['Forecasting'])
df.set_index('Forecasting', inplace=True)
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.index.freq = 'MS'
```

```python
df.dtypes
```

```python
series_data = df['Monthly Revenue']
```

```python
model = Holt(series_data, exponential=False, initialization_method='estimated')
fit_model = model.fit(optimized=True)
```

```python
forecast_steps = 12
forecast = fit_model.forecast(steps=forecast_steps)
```

```python
print(forecast)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
```

```python
train_data = series_data.iloc[:-12]
test_data = series_data.iloc[-12:]
```

```python
model = Holt(train_data, exponential=False, initialization_method='estimated')
fit_model = model.fit(optimized=True)
```

```python
predictions = fit_model.forecast(len(test_data))
```

```python
rmse = np.sqrt(mean_squared_error(test_data, predictions))
mape = mean_absolute_percentage_error(test_data, predictions) * 100
```

```python
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"Mean Absolute Percentage Error (MAPE): {mape:.2f}%")
```
