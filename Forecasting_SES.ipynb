```python
df = pd.read_excel('Project Finance.xlsx', sheet_name='Forecasting')
```

```python
df.head()
```

```python
df['Forecasting'] = pd.to_datetime(df['Forecasting'], errors='coerce')
df.set_index('Forecasting', inplace=True)
```

```python
df.head()
```

```python
df = df.asfreq('MS')
```

```python
df.head()
```

```python
target_data = df['Monthly Revenue']
```

```python
model = SimpleExpSmoothing(target_data, initialization_method='estimated')
model_fit = model.fit(optimized=True)
```

```python
print(f"Nilai Alpha optimal: {model_fit.model.params['smoothing_level']}")
```

```python
forecast_resault = model_fit.forecast(steps=12)
print(forecast_resault)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
```

```python
jumlah_data = 12
train_data = target_data.iloc[:-jumlah_data]
test_data = target_data.iloc[-jumlah_data:]
```

```python
model = SimpleExpSmoothing(train_data, initialization_method='estimated')
model_fit = model.fit(optimized=True)
```

```python
predictions = model_fit.forecast(steps=len(test_data))
```

```python
rmse = np.sqrt(mean_squared_error(test_data, predictions))
mape = mean_absolute_percentage_error(test_data, predictions)*100
```

```python
print(f"Nilai RMSE: {rmse:.2f}")
print(f"Nilai MAPE: {mape:.2f}%")
```
